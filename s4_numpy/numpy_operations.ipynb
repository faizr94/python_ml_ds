Let's create a 2 dimensional array

```python
arr_2d = np.arange(0,25).reshape(5,5)
arr_2d
```

```python
arr_2d.shape
```

```python
arr_2d.sum()
```

```python
# the specified axis is the dimension that gets collapsed or eliminated from the array's shape
# so if axis = 0, since 0 represents rows so we are summing down the cols
# conversely, if axis = 1, then we are summing across the rows
col_sum = arr_2d.sum(axis=0)
row_sum = arr_2d.sum(axis=1)
```

```python
col_sum, row_sum
```
