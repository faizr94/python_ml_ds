Numpy Indexing and Selection

```python
import numpy as np
```

```python
arr = np.arange(0,11)
```

```python
arr
```

```python
arr[8]
```

```python
arr[1:5]
```

```python
arr[:5]
```

```python
arr[5:]
```

```python
# reassignment
arr[0:5] = 100
arr
```

```python
arr = np.arange(0,11)
```

slicing a section of an array and set it to a new variable

```python
slice_of_arr = arr[0:5]
slice_of_arr
```

```python
arr
```

```python
slice_of_arr[:] = 99
slice_of_arr
```

```python
arr
```

Indexing on 2d arrays or matrixes

```python
arr_2d = np.array([[5,10,15], [20,25,30], [35,40,35]])
arr_2d
```

```python
arr_2d.shape
```

```python
# first row
arr_2d[0]
```

```python
# last row
arr_2d[2]
```

```python
# what about grabbing a single element? number 25 for example
arr_2d[1][1]
```

```python
# I could also do this
arr_2d[1,1]
```

```python
# lets say we wanted to slice 10,15 and 25,30
# :2 → rows 0 and 1 (because we started from index zero and up to but not including 2)
# 1: → columns starting from index 1 to the end
arr_2d[:2,1:]
```
