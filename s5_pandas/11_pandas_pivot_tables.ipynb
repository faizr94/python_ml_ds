```python
import pandas as pd
import numpy as np
```

```python
csv_file = "./data/Sales_Funnel_CRM.csv"
df = pd.read_csv(csv_file)
```

```python
# We use these 3 cols first, company, product and licenses
licenses = df[['Company', 'Product', 'Licenses']]
```

```python
pd.pivot(data=licenses, index='Company', columns='Product', values='Licenses')
```

```python
# Doing some aggregations. Get the total sum of licenses and sale price for each company 

pd.pivot_table(data=df[['Licenses', 'Sale Price', 'Company']], index='Company', aggfunc='sum')
```

```python
# Technically we can just use a groupby
df.groupby(by='Company')[['Licenses', 'Sale Price']].sum()
```

```python
# Using 2 cols for index
pd.pivot_table(data=df, index=['Account Manager', 'Contact'], values='Sale Price', aggfunc='sum')
```

```python
pd.pivot_table(data=df, index=['Account Manager', 'Contact', 'Product'], values='Sale Price', aggfunc='sum', fill_value=0, margins=True)
```
