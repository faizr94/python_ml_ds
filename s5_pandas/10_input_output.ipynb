### 3) SQL

- we will use SQLite for this case since it already comes with Pyhthon

```python
# Create our temporary DB first 
from sqlalchemy import create_engine
```

```python
temp_db = create_engine("sqlite:///:memory:")
```

```python
df = pd.DataFrame(data=np.random.randint(low=0,high=100,size=(4,4)), columns=['a','b','c','d'])
```

```python
df
```

```python
df.to_sql(
    name="new_table",
    con=temp_db
)
```

```python
# Reading frm the SQL db
new_df = pd.read_sql(
    sql="new_table",
    con=temp_db
)
```

```python
new_df
```

```python
# Passing a query to pandas and returns result as a dataframe
pd.read_sql_query(
    sql = "SELECT a,c FROM new_table",
    con=temp_db
)
```
