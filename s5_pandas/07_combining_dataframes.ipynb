### 1) Concatenation

- if both sources are already in the same format, then a concatenation using the
pd.concat() call is sufficient
- we can perform the concatenation on either the rows or cols 

```python
import numpy as np
import pandas as pd
```

```python
data_one = {
    "A": ["A0", "A1", "A2", "A3"],
    "B": ["B0", "B1", "B2", "B3"]
}

data_two = {
    "C": ["C0", "C1", "C2", "C3"],
    "D": ["D0", "D1", "D2", "D3"]
}
```

```python
one = pd.DataFrame(data_one)
two = pd.DataFrame(data_two)
```

```python
one.head()
```

```python
two.head()
```

```python
# specify axis=1 to join along the cols
pd.concat([one, two], axis=1)
```

```python
# joining along the rows, similar to a union in SQL
pd.concat([one, two], axis=0)
```

```python
# if we match up the col names in the two dataframe, we can avoid the NaNs as in the above 
two.columns = one.columns
pd.concat([one, two], axis=0)
```

### 2) Inner Merge

- similar to an inner join in SQL
- the .merge() method takes in a key argument labelled how
- like in SQL, we can use either inner, outer, left or right

```python
registrations = pd.DataFrame({'reg_id':[1,2,3,4],'name':['Andrew','Bobo','Claire','David']})
logins = pd.DataFrame({'log_id':[1,2,3,4],'name':['Xavier','Andrew','Yolanda','Bobo']})
```

```python
registrations
```

```python
logins
```

```python
registrations.merge(right=logins, how='inner', on="name")
```

### 3) Left and Right Merge

```python
registrations.merge(right=logins, how="left", on="name")
```

```python
registrations.merge(right=logins, how="right", on="name")
```

### 4) Outer Merge

- similar to a full join

```python
registrations.merge(right=logins, how="outer", on="name")
```

### Joining on an index instead of a col

```python
registrations = registrations.set_index("name")
registrations
```

```python
logins
```

```python
registrations.merge(right=logins, left_index=True, right_on="name", how="inner")
```

```python
# dealing with different key column names in the join tables 
registrations = registrations.reset_index()
```

```python
registrations.columns = ['reg_name', 'reg_id']
registrations
```

```python
registrations.merge(right=logins, how="inner", left_on="reg_name", right_on="name")
```

Adding or tagging duplicate cols

```python
registrations.columns = ['name', 'id']
logins.columns = ['id', 'name']
```

```python
registrations
```

```python
logins
```

```python
registrations.merge(right=logins, how="inner", on="name")
```
