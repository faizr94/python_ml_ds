### GroupBy Operations - Part One

- a groupby() operation allows us to examine data on a per category basis

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("data/mpg.csv")
df.head()
```

```python
df['model_year'].unique()
```

```python
# let's see the avg of numerical cols grouped by model year
df.groupby(by='model_year').mean(numeric_only=True)
```

```python
df.groupby(by='model_year')["mpg"].mean()
```

```python
# group by list of cols too 
df.groupby(by=['model_year', 'cylinders']).mean(numeric_only=True).head()
```

```python
# we can also apply describe on a groupby
df.groupby(by="model_year").describe()
```

```python
year_cyl = df.groupby(by=["model_year", "cylinders"]).mean(numeric_only=True)
year_cyl.head()
```

```python
year_cyl.index.names
```

```python
year_cyl.index.levels
```

```python
# lets say we only wanted to get records where model_year = 70
year_cyl.loc[70]
```

```python
year_cyl.loc[[70, 82]]
```

```python
# grabbing 8 cylinder cars for the year 1970
year_cyl.loc[(70, 8)]
```

### Groupby Operations - Part Two

```python
# when using xs(cross-section), it only takes in a single key as the value
year_cyl.xs(key=70, level="model_year")
```

- we can use cross-section if we wanted to grab data based on the inside index 
- example; we wanted to grab only 4 cylinder cars but for every single year 

```python
year_cyl.xs(key=4, level="cylinders")
```

```python
# swapping the levels 
year_cyl.swaplevel()
```

```python
# sort multilevel index 
year_cyl.sort_index(level="model_year", ascending=False).head()
```

```python
# customise aggregate functions per category
#  but we need to choose the numeric cols first
df.select_dtypes('number').agg(['std', 'mean'])
```

```python
df.agg({'mpg': ['max', 'mean'], 'weight': ['mean', 'std']})
```
