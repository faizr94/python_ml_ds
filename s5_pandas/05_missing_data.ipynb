```python
# Exploring a dataset with null vals inside 
df = pd.read_csv("data/movie_scores.csv")
df.head()
```

```python
# Checking and selecting null values 
df.isnull()
```

```python
# Filtering rows where the pre_movie_score is not null
df[df['pre_movie_score'].notnull()]
```

```python
# What if we're looking for an actor where the pre_movie_score is missing but first_name is not null
df[(df['pre_movie_score'].isnull()) & (df['first_name'].notnull())]
```

### Dropping missing data

```python
# Dropping Data
# thresh=1 means that as long as the row has a non-null value, it will still be kept
df.dropna(thresh=1)
```

```python
# Axis arguments 
# This would simply drop cols that are missing any values, which is basically all cols
df.dropna(axis=1)
```

```python
# the default would be:
df.dropna(axis=0)
```

```python
# we can add in the subset argument to define which cols to consider for the missing values 
df.dropna(subset=['last_name'])
```

### Filling in missing data

```python
# this will fill in NEW VALUE for every NaN
df.fillna('NEW VALUE')
```

```python
# typically we grab the col we are interested in filling 
df['pre_movie_score'].fillna(0)
```

```python
# what if we wanted to fill in based on the mean?
pre_score_avg = df['pre_movie_score'].mean()
df['pre_movie_score'].fillna(pre_score_avg)
```

### Interpolation

```python
airline_tix = {
    'first': 100,
    'business': np.nan,
    'economy-plus': 50,
    'economy': 30
}

ser = pd.Series(airline_tix)
ser
```

```python
# this does a linear interpolation between 100 and 50, to determine the business class tix price 
ser.interpolate()
```
