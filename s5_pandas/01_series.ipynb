- A series is a data structure in Pandas that holds an array of information along with a 
named index
- the named index is what differentiates it from a simply numpy array
- a Pandas series basically adds on a labelled index

```python
import numpy as np
import pandas as pd
```

```python
myindex = ['USA', 'Canada', 'Mexico']
mydata = [1776, 1867, 1821]
```

```python
myseries = pd.Series(data=mydata, index=myindex)
myseries
```

```python
myseries.iloc[0]
```

creating a series frm a dictionary

```python
ages = {'Sam':5, 'Frank':10, 'Spike':7}
```

```python
pd.Series(ages)
```

```python
# Imaginary sales data for Q1 and Q2 of a company
q1 = {'Japan': 80, 'China': 450, 'India':200, 'USA':250}
q2 = {'Brazil': 100, 'China': 5000, 'India':210, 'USA':260}
```

```python
sales_q1 = pd.Series(q1)
sales_q2 = pd.Series(q2)
```

```python
sales_q1
```

```python
sales_q2
```

```python
sales_q1['Japan']
```

```python
# View keys available in a series
sales_q1.keys()
```

Operations between 2 series

```python
sales_q1 *2
```

```python
sales_q1.add(sales_q2,fill_value=0)
```
