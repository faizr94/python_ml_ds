### Part 3 - Working with Columns

```python
# Getting single cols
df['total_bill'].head()
```

```python
# Getting multiple cols
df[['total_bill','tip']].head()
```

```python
# Lets say we wanted to add a new col which is basically the tip % of the total bill
# Take note that if you create a new col with an existing col name, it will essentially 
# overwrite the current data
df['tip_percentage'] = np.round((df['tip']/df['total_bill'])*100,2)
```

```python
df.head()
```

```python
# Removing cols
# Take note that this change is not permanent unless u specify either inplace=True
# Or we do reassignment
df.drop(
    'tip_percentage',
    axis=1
)
```

```python
df = df.drop(
    'tip_percentage',
    axis=1
)
df.head()
```

### Part 4 - Working with Rows

```python
# Payment ID now becomes our unique index
# once again, we need to reassign back to the df to make it fixed
df = df.set_index('Payment ID')
df.head()
```

```python
# To reset the index
df = df.reset_index()
df.head()
```

```python
# using index
df.iloc[0]
```

```python
# using the labelled rows
df.loc['Sun2959']
```

```python
# for multiple rows, can use slice notation 
df.iloc[0:4]
```

```python
# alternative method
df.loc[['Sun2959', 'Sun4608']]
```

```python
# this change is not permanent
df.drop(
    'Sun2959',
    axis=0
)
```

Note that df.append has already been deprecated

```python
# Appending a row to your dataframe
one_row = df.iloc[[0]]
one_row
```
