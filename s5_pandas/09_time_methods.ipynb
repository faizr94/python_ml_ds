```python
sales = pd.read_csv("./data/RetailSales_BeerWineLiquor.csv", parse_dates=[0])
sales.info()
```

```python
# Resampling 
sales.set_index("DATE", inplace=True)
sales.head()
```

```python
# This groups everything by year and get the mean for each year
sales.resample(rule='YE').mean()
```
