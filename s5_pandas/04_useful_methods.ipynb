### 1. Useful Pandas Methods on single columns 

#### 1.1 Apply Method on Single Col

- the .apply() method can be used to apply any custom python function to each row in a series 

```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv('data/tips.csv')
df.info()
```

```python
# Let's say we wanted to get the last 4 digits from CC Number 
# However, since its an int, we cannot use index notation so it needs to be converted to a str first 
# We can create our own function first and then apply it 

def last_four(s):
    return str(s)[-4:]

df['CC Number'].apply(last_four)
```

```python
# Let's check out the total bill col now
# Let's create a custom function that takes in a single price and then returns either $, $$ or $$$ to indicate how pricey it was 
df['total_bill'].head()
```

```python
def yelp(price):
    if price < 10:
        return "$"
    elif price >= 10 and price < 30:
        return "$$"
    else:
        return "$$$"
```

```python
# Applying the function to total_bill
df['total_bill'].apply(yelp)
```

### 2. Useful Pandas Methods on multiple columns 

### 2.1 Quick review on lambda functions 
- Since lambda functions are really just anonymous/single use functions, we can do the following:

```python
# Named function
def simple(num):
    return num*2
```

```python
simple(2)
```

```python
# Converted named function to a lambda expression
lambda num: num*2
```

### 2.2 Applying a single lambda expression to a column

```python
df['total_bill'].apply(lambda num: num*2)
```

### 2.3 Using apply that uses multiple columns

```python
df.head()
```

```python
# Let's use the apply method on the total_bill col and the tip col
# We will then label a tip as generous or other, depending on the ratio of the tip to the total bill

def quality(total_bill, tip):
    if tip/total_bill > 0.1:
        return "Generous"
    else:
        return "Other"
```

```python
# check function results 
quality(16.99, 1.01)
```

```python
df[['total_bill', 'tip']].apply(lambda df: quality(df['total_bill'], df['tip']),axis=1)
```

### 2.4 Using np.vectorize

```python
np.vectorize(quality)(df['total_bill'], df['tip'])
```

- The previous df.apply() is slower in this case because for each row, Pandas has to build an actual Series object from that row's values before calling the function 
- This needs to be done for every single iteration 

- np.vectorize is much faster because it uses df['total_bill'] and df['tip'] as numpy arrays upfront
