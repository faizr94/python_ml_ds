- conditional filtering allows us to select rows based on a condition on a column

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('data/tips.csv')
df.head()
```

### Conditional filtering based on 1 condition

```python
# get records where total_bill > 40
# 2 step method 
bool_series = df['total_bill'] > 40
bool_series
```

```python
df[bool_series].head()
```

```python
# 1 step method
df[df['total_bill'] > 40].head()
```

```python
# All rows where sex is male
df[df['sex'] == "Male"].head()
```

### Conditional filtering based on multiple conditions

- using the AND & operator 
- using the OR | operator

-> Step by step

```python
df['total_bill'] > 30 
```

```python
df[(df['total_bill'] > 30) & (df['sex'] == "Male")].head()
```

```python
df[(df['total_bill'] > 30) | (df['sex'] == "Male")].head()
```

```python
# Filter records where it is the weekend 
# df[(df['day'] == 'Sun') | (df['day'] == 'Sat')].head()
```

```python
# faster method for conditional filtering when you are filtering on the same col
options = ['Fri','Sun']
df[df['day'].isin(options)].head()
```
